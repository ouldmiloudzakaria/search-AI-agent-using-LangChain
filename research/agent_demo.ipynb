```python
import os 
import certifi
from dotenv import load_dotenv

from langchain_groq import ChatGroq
from langchain_community.tools.tavily_search import TavilySearchResults 
from langchain import hub
from langchain.tools import tool
import requests
```

```python
from langchain.agents import create_react_agent, AgentExecutor  # using for creating the agent and executing it
```

```python
#LOAD ENVIRONMENT VARIABLES
os.environ["SSL_CERT_FILE"] = certifi.where() #use this to set the SSL certificate file path for secure connections
load_dotenv() # use this to load the environment variables from the .env file

GROQ_API_KEY = os.getenv("GROQ_API_KEY")    
TAVILY_API_KEY = os.getenv("TAVILY_API_KEY")
```

```python
search_tool = TavilySearchResults(max_results=3) # Initialize the TavilySearchResults tool with a maximum of 3 results and the API key
```

```python
result = search_tool.run("What is the capital of France?") # Use the run method to perform a search query and get the results
result 
```

```python
# llm
llm = ChatGroq(
    api_key=os.getenv("GROQ_API_KEY"),  # Use the API key from the environment variable
    temperature=0,  # Set the temperature for response variability 
    model="qwen/qwen3.8-27b",  # Specify the model to use        
) 
```

```python
response = llm.invoke("What is the capital of France?")  
response
```

```python
# prompt
prompt = hub.pull("hwchase17/react")
```

```python
# tools
tools = [search_tool]  # Create a list of tools to be used by the agent
```

```python
#create the agent
agent = create_react_agent(llm=llm, tools=tools, prompt=prompt)
```

```python
# execute the agent 
agent_executor = AgentExecutor(agent=agent, tools=tools ,verbose=True)  # verbose mean that it will print the intermediate steps and actions taken by the agent during execution
```

```python
# run 

respons = agent_executor.invoke({
    "input":(
        "tell me the lastest news about iran and usa"
        )
})  
```

```python
print(respons["output"])  # Print the final output of the agent's execution
```
